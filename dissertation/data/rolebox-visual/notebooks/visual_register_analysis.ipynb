# Visual Register Analysis: EIT Knowledge and Innovation Communities

This notebook demonstrates the visual register analysis pipeline for analyzing
how EIT Communities visually construct and communicate institutional identity.

## Theoretical Framework

Following Jancsary et al. (2018), we analyze visual registers through three metafunctions:

1. **Ideational**: Content of representation (participants, processes, settings)
2. **Interpersonal**: Viewer positioning (contact, distance, angle, coding orientation)
3. **Textual**: Composition and framing

```python
import json
import sys
sys.path.insert(0, '../src')

from rolebox_visual import (
    VisualCoding, VisualCorpus,
    GazeDetector, VisualRegisterAnalyzer,
    CooccurrenceNetwork, GazeClusterer,
)
from rolebox_visual.coding import (
    ParticipantType, ProcessType, SettingType,
    ContactLevel, SocialDistance, VerticalAngle,
    CodingOrientation,
)
```

## 1. Load Analysis Results

Load the pre-computed gaze analysis from the full corpus (n=9,912 visuals).

```python
# Load gaze analysis results
with open('../data/outputs/gaze_analysis.json', 'r') as f:
    gaze_results = json.load(f)

print(f"Analyzed {gaze_results['metadata']['n_visuals_analyzed']:,} visuals")
print(f"Across {gaze_results['metadata']['n_kics']} KICs")
print(f"\nDetected {len(gaze_results['gazes'])} characteristic gazes")
```

## 2. Corpus-Level Statistics

Examine the distribution of visual elements across the corpus.

```python
stats = gaze_results['corpus_statistics']

print("=== Visual Type Distribution ===")
for vtype, count in stats['by_type'].items():
    pct = count / stats['n_visuals'] * 100
    print(f"  {vtype:12s}: {count:5d} ({pct:5.1f}%)")

print("\n=== Vertical Angle Distribution ===")
for angle, pct in stats['angle_distribution'].items():
    print(f"  {angle:20s}: {pct*100:5.1f}%")
    
print("\n=== Key Finding: Egalitarian Dominance ===")
print(f"  Eye-level angles: {stats['angle_distribution']['equality']*100:.1f}%")
print("  → EIT constructs viewers as PEERS, not subordinates or authorities")
```

## 3. The Three Gazes

Examine the characteristic visual configurations identified through clustering.

```python
print("=== Characteristic Gazes ===")
print("="*60)

for gaze in gaze_results['gazes']:
    print(f"\n{gaze['name'].upper()}")
    print(f"Prevalence: {gaze['prevalence']*100:.0f}% ({gaze['n_visuals']:,} visuals)")
    print(f"Description: {gaze['description']}")
    print(f"\nDominant participants: {', '.join(gaze['dominant_participants'][:3])}")
    print(f"Dominant processes: {', '.join(gaze['dominant_processes'][:3])}")
    print(f"Dominant settings: {', '.join(gaze['dominant_settings'][:3])}")
    print(f"\nVerbal topics: {', '.join(gaze['verbal_topics'])}")
    print("-"*60)
```

## 4. KIC Profiles: Gaze Distribution

How do different KICs emphasize different gazes?

```python
print("=== Gaze Profiles by KIC ===")
print(f"{'KIC':<20} {'Partnership':>12} {'Impact':>12} {'Talent':>12} {'Primary':>12}")
print("="*70)

for kic, profile in gaze_results['kic_profiles'].items():
    gp = profile['gaze_profile']
    print(f"{kic:<20} {gp['partnership']*100:>11.0f}% {gp['impact']*100:>11.0f}% {gp['talent']*100:>11.0f}% {profile['primary_gaze']:>12}")
```

## 5. Visual-Verbal Divergence

A key finding: visual and verbal registers often diverge, revealing taken-for-granted assumptions.

```python
print("=== Visual-Verbal Divergence Indicators ===")
print("="*70)

for dimension, data in gaze_results['divergence_indicators'].items():
    print(f"\n{dimension.upper()}")
    print(f"  Divergence score: {data['divergence_score']:+.3f}")
    print(f"  Interpretation: {data['interpretation']}")
```

## 6. Co-occurrence Network

Load and examine the co-occurrence network structure.

```python
with open('../data/outputs/cooccurrence_network.json', 'r') as f:
    network_data = json.load(f)

print("=== Network Statistics ===")
print(f"Nodes: {network_data['network']['n_nodes']}")
print(f"Edges: {network_data['network']['n_edges']}")
print(f"Density: {network_data['network']['density']:.2f}")
print(f"Modularity: {network_data['network']['modularity']:.2f}")

print("\n=== Detected Communities ===")
for comm_id, comm in network_data['communities'].items():
    print(f"\nCommunity {comm_id}: {comm['name']}")
    print(f"  Nodes: {comm['n_nodes']}")
    print(f"  Key elements: {', '.join(comm['key_elements'][:5])}")
    print(f"  Interpretation: {comm['interpretation']}")
```

## 8. Coding Orientation by KIC

How do coding orientations (truth criteria) vary across KICs?

```python
print("=== Coding Orientation by KIC ===")
print(f"{'KIC':<20} {'Natural':>10} {'Sensory':>10} {'Tech':>10} {'Abstract':>10}")
print("="*62)

for kic, profile in gaze_results['kic_profiles'].items():
    od = profile['orientation_distribution']
    print(f"{kic:<20} {od['naturalistic']*100:>9.0f}% {od['sensory']*100:>9.0f}% {od['technological']*100:>9.0f}% {od['abstract']*100:>9.0f}%")

print("\n→ Industrial KICs (Manufacturing, RawMaterials) show higher technological orientation")
print("→ Human-centered KICs (Food, Culture) show higher naturalistic orientation")
```
